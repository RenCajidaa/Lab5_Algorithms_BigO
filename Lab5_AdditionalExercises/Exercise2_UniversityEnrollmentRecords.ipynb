```python
LAST_NAME = "Cajida"
STUDENT_ID = "TUPM-26-1243"
FAVORITE_ARTIST = "TUKI."

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(c) - 64 for c in FAVORITE_ARTIST.upper() if c.isalpha())

print("Seed:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Name Length:", NAME_LENGTH)
print("Artist Length:", ARTIST_LENGTH)
print("Artist Sum:", ARTIST_SUM)
```

```python
dept_a = [SEED_DIGIT * 10 + i * ID_SUM for i in range(5)]
dept_b = [ARTIST_LENGTH * 10 + i * NAME_LENGTH for i in range(5)]

def merge_sorted(a, b):
    merged, i, j, comparisons = [], 0, 0, 0
    while i < len(a) and j < len(b):
        comparisons += 1
        if a[i] <= b[j]:
            merged.append(a[i])
            i += 1
        else:
            merged.append(b[j])
            j += 1
    merged.extend(a[i:])
    merged.extend(b[j:])
    return merged, comparisons

merged, comparisons = merge_sorted(dept_a, dept_b)

print("Department A:", dept_a)
print("Department B:", dept_b)
print("Consolidated:", merged)
print("Records before:", len(dept_a), "+", len(dept_b), "=", len(dept_a) + len(dept_b))
print("Records after:", len(merged))
print("Comparisons:", comparisons)
print("Ordered correctly:", merged == sorted(merged))
print("Time Complexity: O(n + m)")
print("Space Complexity: O(n + m)")
```

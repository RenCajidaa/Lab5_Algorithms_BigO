```python
LAST_NAME = "Cajida"
STUDENT_ID = "TUPM-26-1243"
FAVORITE_ARTIST = "TUKI."

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(c) - 64 for c in FAVORITE_ARTIST.upper() if c.isalpha())

print("Seed:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Name Length:", NAME_LENGTH)
print("Artist Length:", ARTIST_LENGTH)
print("Artist Sum:", ARTIST_SUM)
```

```python
import time

def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps

def selection_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps

def insertion_sort(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts

def measure(sort_function, data, trials=5):
    times = []
    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        times.append(time.perf_counter() - start)
    return result, sum(times) / trials
```

```python
orders = [
    (ID_SUM * 3) % 100, (SEED_DIGIT * 11) % 100,
    (NAME_LENGTH * 7) % 100, 42, 17, 89, 31, 56
]

conditions = {
    "Sorted": sorted(orders),
    "Reverse": sorted(orders, reverse=True),
    "Random": orders
}

print("Generated Orders:", orders)

for condition, data in conditions.items():
    print(f"\n{condition} input: {data}")
    print(f"{'Algorithm':<18}{'Comparisons':>12}{'Movements':>12}{'Time (s)':>14}{'Verified':>10}")
    for name, func in (("Bubble Sort", bubble_sort),
                       ("Selection Sort", selection_sort),
                       ("Insertion Sort", insertion_sort)):
        result, elapsed = measure(func, data)
        print(f"{name:<18}{result[1]:>12}{result[2]:>12}{elapsed:>14.8f}{str(result[0] == sorted(data)):>10}")
```
